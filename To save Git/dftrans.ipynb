```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv('/Users/Tudor.Orosz/Documents/School/Thesis/Coding in Python/dataset_marketing2.csv')
df2 = pd.read_csv('/Users/Tudor.Orosz/Documents/School/Thesis/Coding in Python/long.csv')
```

```python
df1
```

```python
df2 = df2.fillna(0)
df2
```

```python
df2['Column name'] = df2['Column name'].str.replace('^M_', '', regex=True)
df2
```

```python
df2 = df2.T

df2
```

```python
df2 = df2.set_axis(df2.iloc[0], axis=1).drop(df2.index[0])

df2
```

```python
# calculate the sum_M column
df1['sum_M'] = np.sum(df1[df2.columns] * df2.values[0], axis=1)
```

```python
df1
```

```python
# create new column marketing_mql with initial value 0
df1['marketing_mql'] = 0

# set value to 1 for rows where sum_M is greater than or equal to 50
df1.loc[df1['sum_M'] >= 50, 'marketing_mql'] = 1
```

```python
df1
```

```python
from sklearn.metrics import confusion_matrix

# Create the confusion matrix
cm = confusion_matrix(df1['opportunityid'], df1['marketing_mql'])

# Print the confusion matrix
print(cm)
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# True values
y_true = df1['opportunityid']

# Predicted values
y_pred = df1['marketing_mql']

# Create confusion matrix
cm = confusion_matrix(y_true, y_pred)

# Create labels
labels = ['Not MQL', 'MQL']

# Plot confusion matrix
ax = plt.subplot()
sns.heatmap(cm, annot=True, ax=ax, cmap='Blues', fmt='g')

# Add labels to the plot
ax.set_xlabel('Predicted labels')
ax.set_ylabel('True labels')
ax.set_title('Confusion Matrix')
ax.xaxis.set_ticklabels(labels)
ax.yaxis.set_ticklabels(labels)
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, precision_score, recall_score, fbeta_score, roc_auc_score

# True values
y_true = df1['opportunityid']

# Predicted values
y_pred = df1['marketing_mql']

# Create confusion matrix
cm = confusion_matrix(y_true, y_pred)

# Calculate recall
recall = recall_score(y_true, y_pred)

# Calculate precision
precision = precision_score(y_true, y_pred)

# Calculate F2-score
f2_score = fbeta_score(y_true, y_pred, beta=2)

# Calculate AUC score
auc_score = roc_auc_score(y_true, y_pred)

# Print the scores
print(f"Recall: {recall:.2f}")
print(f"Precision: {precision:.2f}")
print(f"F2-score: {f2_score:.2f}")
print(f"AUC score: {auc_score:.2f}")
```
